## The Theoretical Appeal

Airy beams are a specialized class of wavepackets defined by the Airy function. For directed energy and remote sensing, they offer three theoretical advantages over standard lasers:

*   **Diffraction Resistance:** Standard Gaussian laser beams naturally widen and lose intensity over long distances. Airy beams are non-diffracting, maintaining a tight focal point over much longer ranges.
*   **Self-Healing:** If the beam's path is partially blocked by atmospheric particulates, smoke, or debris, the interference patterns of the diffracting wavelets automatically reconstruct the beam's main lobe further downstream.
*   **Self-Acceleration:** The peak intensity of an Airy beam propagates along a curved, parabolic trajectory, effectively "bending" without the use of physical mirrors or lenses.

For missile defense, atmospheric turbulence and thermal blooming are massive hurdles that distort and weaken lasers before they can destroy a distant target. The resilience of Airy beams means they could theoretically punch through these atmospheric distortions much more effectively.

## The Engineering Reality

Despite the potential, physics and materials science currently prevent Airy beams from being weaponized at the scale required for NMD:

*   **The Infinite Energy Problem:** A mathematically "true" Airy beam requires an infinite amount of energy and an infinite transverse spatial spread. 
*   **Truncation Limits:** To generate them in the real world, engineers must create "truncated" (finite-energy) Airy beams using spatial light modulators or phase masks. However, truncated beams only maintain their self-healing and non-diffracting properties for a limited distance before eventually dispersing like normal light.
*   **Megawatt Power Scaling:** Modulating a standard laser into an Airy profile involves delicate optical components. Scaling these metasurfaces or phase masks to survive the hundreds of kilowatts or megawatts of continuous power required to neutralize a ballistic missile is currently unfeasible; the optics would melt or destroy themselves under the thermal load.

## What NMD Uses Instead

Instead of exotic wavepackets, current tactical directed-energy programs and NMD research rely on high-energy fiber or solid-state lasers paired with **adaptive optics**. Rather than altering the fundamental wave structure, adaptive optics use sensors and rapidly deformable mirrors to measure atmospheric turbulence in real-time. The system then pre-distorts a standard Gaussian beam so that the atmosphere naturally corrects the beam and focuses it precisely onto the target.

This module uses `anise::almanac::Almanac` to compute the expected geometric state of a target body relative to a ground station (or the Earth center). It then injects a simulated anomaly (e.g., a localized fluidic drag or timing delay) to generate the "observed" state and calculates the $L_2$ norm of the residual vector.

```rust
use anise::almanac::Almanac;
use anise::errors::AlmanacError;
use anise::time::Epoch;
use nalgebra::Vector3;

/// A simulator for calculating ephemeris targets and comparing them
/// against perturbed (anomalous) ground observations.
pub struct ObservationSimulator<'a> {
    almanac: &'a Almanac,
    observer_frame: usize, // NAIF ID for observer (e.g., 399 for Earth)
    target_frame: usize,   // NAIF ID for target (e.g., 199 for Mercury)
}

impl<'a> ObservationSimulator<'a> {
    pub fn new(almanac: &'a Almanac, observer_frame: usize, target_frame: usize) -> Self {
        Self {
            almanac,
            observer_frame,
            target_frame,
        }
    }

    /// Computes the exact ephemeris state vector from the loaded SPK/BSP files.
    pub fn compute_expected_position(&self, epoch: Epoch) -> Result<Vector3<f64>, AlmanacError> {
        // Retrieve the translational state of the target relative to the observer
        let state = self.almanac.translate(
            self.observer_frame,
            self.target_frame,
            epoch,
            None, // No aberration corrections applied for geometric state
        )?;

        // Return the position vector (X, Y, Z in kilometers)
        Ok(state.position)
    }

    /// Simulates a physical observation by taking the ground-truth ephemeris
    /// and injecting a specific physical perturbation vector. 
    pub fn simulate_observed_position(
        &self, 
        epoch: Epoch, 
        perturbation_km: Vector3<f64>
    ) -> Result<Vector3<f64>, AlmanacError> {
        let true_position = self.compute_expected_position(epoch)?;

        // Inject the unmodeled anomaly (e.g., dilatant fluid drag shift)
        let observed_position = true_position + perturbation_km;

        Ok(observed_position)
    }

    /// Measures the absolute scalar error between the expected model and the observed data.
    pub fn measure_residual_error(
        &self, 
        epoch: Epoch, 
        perturbation_km: Vector3<f64>
    ) -> Result<f64, AlmanacError> {
        let expected = self.compute_expected_position(epoch)?;
        let observed = self.simulate_observed_position(epoch, perturbation_km)?;

        // Calculate the residual vector and return its L2 norm (magnitude)
        let residual = observed - expected;
        Ok(residual.norm())
    }
}
```

### 3. The Genetic Optimizer (`src/ga.rs`)
This module manages the population and drives the evolutionary process. It evaluates fitness based on the sum of residual errors across a batch of observation epochs.

```rust
use rand::seq::SliceRandom;
use rand::thread_rng;
use crate::physics::Genome;

pub struct GeneticOptimizer {
    pub population_size: usize,
    pub mutation_rate: f64,
    pub tournament_size: usize,
}

impl GeneticOptimizer {
    pub fn new(population_size: usize, mutation_rate: f64) -> Self {
        Self {
            population_size,
            mutation_rate,
            tournament_size: 3,
        }
    }

    /// Runs the GA for a set number of generations.
    /// The `fitness_fn` closure accepts a Genome and returns the total residual error (lower is better).
    pub fn optimize<F>(&self, generations: usize, fitness_fn: F) -> Genome
    where
        F: Fn(&Genome) -> f64,
    {
        let mut rng = thread_rng();

        // Initialize random population
        let mut population: Vec<Genome> = (0..self.population_size)
            .map(|_| Genome::random(&mut rng))
            .collect();

        for _gen in 0..generations {
            // 1. Evaluate fitness (We map Genome to Error)
            let mut fitness_scores: Vec<(Genome, f64)> = population
                .iter()
                .map(|g| (*g, fitness_fn(g)))
                .collect();

            // Sort by error ascending (lowest error is the best)
            fitness_scores.sort_by(|a, b| a.1.partial_cmp(&b.1).unwrap());

            // 2. Elitism: Keep the best 10% automatically
            let elite_count = (self.population_size as f64 * 0.1) as usize;
            let mut next_generation: Vec<Genome> = fitness_scores
                .iter()
                .take(elite_count)
                .map(|(g, _)| *g)
                .collect();

            // 3. Selection, Crossover, and Mutation
            while next_generation.len() < self.population_size {
                let parent1 = self.tournament_selection(&fitness_scores);
                let parent2 = self.tournament_selection(&fitness_scores);

                let mut child = parent1.crossover(&parent2, &mut rng);
                child.mutate(self.mutation_rate, &mut rng);

                next_generation.push(child);
            }

            population = next_generation;
        }

        // Return the best genome from the final population
        population
            .into_iter()
            .min_by(|a, b| fitness_fn(a).partial_cmp(&fitness_fn(b)).unwrap())
            .unwrap()
    }

    /// Selects the best individual from a random subset of the population.
    fn tournament_selection(&self, fitness_scores: &[(Genome, f64)]) -> Genome {
        let mut rng = thread_rng();
        let tournament: Vec<&(Genome, f64)> = fitness_scores
            .choose_multiple(&mut rng, self.tournament_size)
            .collect();

        // Return the genome with the lowest error in the tournament
        tournament.into_iter().min_by(|a, b| a.1.partial_cmp(&b.1).unwrap()).unwrap().0
    }
}
```
