The water content in industrial-grade citric and phytic acids varies significantly due to their chemical natures—one is a crystalline solid that likes to hold onto a specific amount of water, while the other is a syrupy, hygroscopic beast that is almost always sold as a solution.

### **Citric Acid ($C_6H_8O_7$)**

Industrial citric acid typically comes in two forms: **Anhydrous** (water-free) and **Monohydrate**.

* **Water Content:**
    * **Monohydrate:** Contains one molecule of water for every molecule of citric acid. This calculates to approximately **8.58% water** by weight.
    * **Anhydrous:** Legally allowed to have trace moisture (usually $<0.5\%$), but for all intents and purposes, it is dry.
* **Efficient Removal:**
    * **Thermal Dehydration:** Heating citric acid monohydrate above **78°C (172°F)** will cause it to lose its crystallization water and convert to the anhydrous form.
    * **Industrial Standard:** A **fluidized bed dryer** or a **rotary dryer** is typically used. In these systems, warm air is passed through the crystals to ensure uniform heating without reaching the melting point (approx. 153°C), which would turn your batch into a sticky mess.

---

### **Phytic Acid ($C_6H_{18}O_{24}P_6$)**

Phytic acid is a different animal. Because it is extremely hygroscopic (it pulls water out of the air like a sponge), it is almost never sold as a pure solid industrially.

* **Water Content:**
    * It is most commonly sold as a **50% (w/w)** or **70% (w/w) aqueous solution**.
    * This means you are dealing with **30% to 50% water**.
* **Efficient Removal:**
    * **Vacuum Evaporation:** This is the most efficient industrial route. Because phytic acid is sensitive to heat and can begin to decompose or "char" at high temperatures, you need to lower the boiling point of water by using a deep vacuum.
    * **Lyophilization (Freeze-Drying):** If you need the phytic acid in a solid or near-solid state without thermal degradation, freeze-drying is the gold standard, though it is more energy-intensive and expensive than vacuum evaporation.
    * **Desiccation:** For small scales, using a vacuum desiccator with a strong drying agent like phosphorus pentoxide ($P_2O_5$) can pull the final traces of water away, though this is slow.

---

### **General Efficiency Tips**

1.  **Vacuum is Your Friend:** For both acids, applying a vacuum allows you to remove water at lower temperatures. This prevents the citric acid from melting and the phytic acid from degrading.
2.  **Surface Area:** If you are drying phytic acid, keep the layer as thin as possible. As it loses water, it becomes incredibly viscous (like thick honey), making it harder for the remaining water molecules to escape the "syrup."
3.  **Temperature Control:** Never exceed **100°C** for phytic acid unless you want to risk hydrolysis and the formation of lower inositol phosphates. For citric acid, staying between **80°C and 100°C** is the "sweet spot" for converting monohydrate to anhydrous efficiently.
