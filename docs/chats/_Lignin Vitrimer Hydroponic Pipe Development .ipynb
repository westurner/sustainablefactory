$$f_s = \frac{St \cdot U}{D} = \frac{0.2 \cdot 30}{0.1} = 60\text{ Hz}$$

$$F_L = \frac{1}{2} \rho U^2 D C_L L$$
$$F_L = 0.5 \cdot 1.225 \cdot (30)^2 \cdot 0.1 \cdot 0.6 \cdot 2 = 33.08\text{ N}$$

## Supercharging the Actuators with Carbon Allotropes

Standard phase-change materials (like paraffin wax) are excellent at expanding, but they are thermal insulators ($\approx 0.2\text{ W/m·K}$). They react sluggishly because the sun's heat takes a long time to penetrate to the core of the actuator. 

By doping the wax with a percolating network of carbon nanomaterials, we can create thermal "highways" that blast the solar pulse instantly through the entire volume, reducing reaction times from minutes to seconds.

### 1. Graphene: The 3D Thermal Scaffold
Instead of mixing loose graphene powder (which clumps), you can use a continuous **Laser-Induced Graphene (LIG)** scaffold or a rhombohedral graphene foam. 
* **The Mechanism:** The liquid wax is drawn into the highly porous graphene sponge via capillary action. Because the graphene matrix is physically continuous, heat doesn't have to jump across polymer boundaries. 
* **The Result:** The thermal conductivity of the composite skyrockets by over 1000%, ensuring the entire wax volume melts and expands simultaneously. 

### 2. Biodegradable CNTs: The High-Speed Arteries
Traditional Carbon Nanotubes (CNTs) are environmentally persistent and pose inhalation risks, making them a poor fit for sustainable infrastructure. However, intentionally modifying them creates a biological kill-switch.
* **The Mechanism:** By heavily functionalizing single-walled CNTs with oxygen-containing groups (creating highly carboxylated ox-SWCNTs), they become susceptible to enzymatic cleavage. 
* **The Loop:** They act as ultra-fast thermal bridges inside the actuator. At the end of the mount's lifecycle, once the polymer matrix degrades, environmental fungi or plant-derived enzymes (like horseradish peroxidase) can attach to the defect sites on the CNTs and fully oxidize them into basic carbon compounds.

### 3. CNF (Carbon Nanofibers): The Structural Lignin Bridge
Carbon nanofibers are thicker and cheaper than CNTs, making them ideal for bulk structural reinforcement of the flexible actuator housing, preventing it from bursting under immense hydraulic pressure.
* **The Mechanism:** To keep this strictly closed-loop, these CNFs can be electrospun directly from kraft lignin—a massive waste stream from the paper and pulp industry. 
* **The Loop:** Upcycling lignin into electrospun CNFs provides a cheap, highly conductive reinforcing mesh that ties into existing industrial waste streams without relying on fossil-derived precursors.

### 4. CNS (Mesoporous Carbon Nanospheres): The Micro-Vaults
A major failure mode of phase-change actuators is the wax slowly leaking out over thousands of diurnal melt/freeze cycles. 
* **The Mechanism:** Mesoporous carbon nanospheres can be used to micro-encapsulate the wax. The wax lives entirely *inside* the microscopic carbon shells.
* **The Result:** The CNS shell provides a massive, highly conductive surface area to absorb the solar pulse, immediately transferring it to the trapped wax inside. The wax melts and expands, pushing against the flexible shell, but it physically cannot leak out, drastically extending the zero-maintenance lifespan of the tracker.
