**Piezoelectric plasma excitation via longitudinal waves** relies on the coupling of mechanical acoustic waves (specifically longitudinal or compression waves) with electrostatic or electromagnetic modes in a medium to generate, sustain, or modulate a plasma state.

This phenomenon spans across two major domains: engineering applications in cold plasma generation, and solid-state plasma physics in semiconductor devices.

* * *

1\. Piezoelectric Direct Discharge (PDD) Plasma

In macro-scale engineering applications, [Piezoelectric Direct Discharge (PDD)](https://en.wikipedia.org/wiki/Piezoelectric_direct_discharge_plasma) is utilized to generate high-efficiency non-equilibrium cold plasmas.

```
[AC Voltage Input] ──> [Piezoelectric Crystal] ──> [Longitudinal Acoustic Wave] ──> [High Electric Field at Tip] ──> [Gas Ionization / Plasma]
```

-   **The Mechanism:** An alternating voltage drives a piezoelectric transformer at its resonant frequency. This input creates strong mechanical **longitudinal waves** (compression and stretching modes) along the crystal length.
-   **The Excitation:** Due to the material's structural polarization, the mechanical stress resulting from the longitudinal acoustic wave multiplies the voltage at the crystal's output tip. This amplification produces a localized electric field high enough (1 to 10 eV) to ionize surrounding air molecules, generating an open-air cold plasma.
-   **Performance Overview:**

    | Metric | Typical Value |
    | --- | --- |
    | **Power Consumption** | ~10 W |
    | **Plasma Generation Efficiency** | Up to 90% |
    | **Neutral Gas Temperature** | < 50 °C |
    | **Electron/Ion Energies** | 1 – 10 eV |
    | **Plasma Density** | 10¹⁴ to 10¹⁶ m⁻³ |

2\. Solid-State Plasma Dynamics in Semiconductors

At the micro- or nanoscale, the interaction occurs internally within **piezoelectric semiconductors** (such as GaAs, GaN, or ZnO).

-   **Acoustoelectric Coupling:** When an acoustic longitudinal wave travels along specific crystal symmetry axes of a piezoelectric semiconductor, it induces a traveling longitudinal electric field via the piezoelectric effect.
-   **Plasma Wave Interaction:** This induced field couples nonlinearly with the solid-state electron plasma density fluctuations.
-   **Parametric Amplification:** If an external DC or RF driving field is applied, the coupling can induce instability or parametric excitation. This can amplify the ultrasonic longitudinal waves or generate distinct longitudinal electron density oscillations within the material’s conduction bands. Recent models also show this framework moving into the terahertz elastic wave domain for ultra-fast acoustic devices.
